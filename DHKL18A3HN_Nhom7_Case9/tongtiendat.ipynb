```python
import numpy as np

sem1 = np.genfromtxt("semester1_scores.csv", delimiter=",", skip_header=1)
sem2 = np.genfromtxt("semester2_scores.csv", delimiter=",", skip_header=1)

sid1 = sem1[:, 0].astype(int)
mid1 = sem1[:, 1]
fin1 = sem1[:, 2]
att1 = sem1[:, 3]

sid2 = sem2[:, 0].astype(int)
mid2 = sem2[:, 1]
fin2 = sem2[:, 2]
att2 = sem2[:, 3]

total1 = 0.3 * mid1 + 0.5 * fin1 + 0.2 * att1
total2 = 0.3 * mid2 + 0.5 * fin2 + 0.2 * att2

diff_scores = sem2[:, 1:] - sem1[:, 1:]

improvement = total2 - total1

idx_top_sem1 = np.argsort(total1)[-10:]
idx_top_sem2 = np.argsort(total2)[-10:]
idx_top_improve = np.argsort(improvement)[-10:]

summary = np.column_stack((sid1, total1, total2, improvement))

print("===== 5 dòng đầu summary =====")
print(summary[:5])

print("\n===== TOP 10 HK1 =====")
print(summary[idx_top_sem1])

print("\n===== TOP 10 HK2 =====")
print(summary[idx_top_sem2])

print("\n===== TOP 10 TIẾN BỘ =====")
print(summary[idx_top_improve])
```

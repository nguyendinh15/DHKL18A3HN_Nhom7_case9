```python
import numpy as np

# 1. Đọc dữ liệu từ 2 file CSV
file_sem1 = "semester1_scores.csv"
file_sem2 = "semester2_scores.csv"

# Dùng genfromtxt để xử lý dòng thiếu/hỏng
data_sem1 = np.genfromtxt(file_sem1, delimiter=",", dtype=None, encoding="utf-8", skip_header=1)
data_sem2 = np.genfromtxt(file_sem2, delimiter=",", dtype=None, encoding="utf-8", skip_header=1)

# 2. Kiểm tra typr, shape
print("Semester 1 dtype:", data_sem1.dtype)
print("Semester 1 shape:", data_sem1.shape)

print("Semester 2 dtype:", data_sem2.dtype)
print("Semester 2 shape:", data_sem2.shape)

# 3. Tách thành các mảng con
# Giả sử file có format:
# sid, midterm, final, attendance_hw

sid1  = data_sem1[:, 0].astype(str)
mid1  = data_sem1[:, 1].astype(float)
final1 = data_sem1[:, 2].astype(float)
att1   = data_sem1[:, 3].astype(float)

sid2  = data_sem2[:, 0].astype(str)
mid2  = data_sem2[:, 1].astype(float)
final2 = data_sem2[:, 2].astype(float)
att2   = data_sem2[:, 3].astype(float)

# 4. Xuất 5 giá trị đầu/cuối
print("\nSemester 1 - 5 dòng đầu:")
print(data_sem1[:5])

print("\nSemester 1 - 5 dòng cuối:")
print(data_sem1[-5:])

print("\nSemester 2 - 5 dòng đầu:")
print(data_sem2[:5])

print("\nSemester 2 - 5 dòng cuối:")
print(data_sem2[-5:])

# 5. Kiểm tra lỗi / thiếu dữ liệu / định dạng sai
def check_data_errors(data, name):
    print(f"\n--- Kiểm tra lỗi dữ liệu {name} ---")

    # Kiểm tra giá trị NaN
    if np.isnan(data[:,1:].astype(float)).any():
        print(" Có giá trị NaN trong bảng.")
    else:
        print(" Không có giá trị NaN.")

    # Điểm phải nằm trong 0–100
    numeric = data[:,1:].astype(float)
    if ((numeric < 0) | (numeric > 100)).any():
        print(" Có điểm nằm ngoài khoảng 0–100.")
    else:
        print(" Tất cả điểm trong khoảng hợp lệ.")

check_data_errors(data_sem1, "Semester 1")
check_data_errors(data_sem2, "Semester 2")

print("\n>>> DỮ LIỆU ĐÃ LÀM SẠCH HOÀN TẤT.")
#
```

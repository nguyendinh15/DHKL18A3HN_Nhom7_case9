```python
import numpy as np

# 1. Đọc dữ liệu từ 2 file CSV
file_sem1 = "semester1_scores.csv"
file_sem2 = "semester2_scores.csv"

# Dùng genfromtxt để xử lý dòng thiếu/hỏng
data_sem1 = np.genfromtxt(file_sem1, delimiter=",", dtype=None, encoding="utf-8", skip_header=1)
data_sem2 = np.genfromtxt(file_sem2, delimiter=",", dtype=None, encoding="utf-8", skip_header=1)

# 2. Kiểm tra typr, shape
print("Semester 1 dtype:", data_sem1.dtype)
print("Semester 1 shape:", data_sem1.shape)

print("Semester 2 dtype:", data_sem2.dtype)
print("Semester 2 shape:", data_sem2.shape)

# 3. Tách thành các mảng con
# Giả sử file có format:
# sid, midterm, final, attendance_hw

sid1  = data_sem1[:, 0].astype(str)
mid1  = data_sem1[:, 1].astype(float)
final1 = data_sem1[:, 2].astype(float)
att1   = data_sem1[:, 3].astype(float)

sid2  = data_sem2[:, 0].astype(str)
mid2  = data_sem2[:, 1].astype(float)
final2 = data_sem2[:, 2].astype(float)
att2   = data_sem2[:, 3].astype(float)

# 4. Xuất 5 giá trị đầu/cuối
print("\nSemester 1 - 5 dòng đầu:")
print(data_sem1[:5])

print("\nSemester 1 - 5 dòng cuối:")
print(data_sem1[-5:])

print("\nSemester 2 - 5 dòng đầu:")
print(data_sem2[:5])

print("\nSemester 2 - 5 dòng cuối:")
print(data_sem2[-5:])

# 5. Kiểm tra lỗi / thiếu dữ liệu / định dạng sai
def check_data_errors(data, name):
    print(f"\n--- Kiểm tra lỗi dữ liệu {name} ---")

    # Kiểm tra giá trị NaN
    if np.isnan(data[:,1:].astype(float)).any():
        print(" Có giá trị NaN trong bảng.")
    else:
        print(" Không có giá trị NaN.")

    # Điểm phải nằm trong 0–100
    numeric = data[:,1:].astype(float)
    if ((numeric < 0) | (numeric > 100)).any():
        print(" Có điểm nằm ngoài khoảng 0–100.")
    else:
        print(" Tất cả điểm trong khoảng hợp lệ.")

check_data_errors(data_sem1, "Semester 1")
check_data_errors(data_sem2, "Semester 2")

print("\n>>> DỮ LIỆU ĐÃ LÀM SẠCH HOÀN TẤT.")
```

```python
import numpy as np

# Đọc dữ liệu từ 2 file
sem1 = np.genfromtxt("semester1_scores.csv", delimiter=",", skip_header=1)
sem2 = np.genfromtxt("semester2_scores.csv", delimiter=",", skip_header=1)

# Tách cột
sid1, mid1, final1, att1 = sem1[:,0], sem1[:,1], sem1[:,2], sem1[:,3]
sid2, mid2, final2, att2 = sem2[:,0], sem2[:,1], sem2[:,2], sem2[:,3]

# Tính điểm tổng hợp cho mỗi kỳ
total1 = 0.3*mid1 + 0.5*final1 + 0.2*att1
total2 = 0.3*mid2 + 0.5*final2 + 0.2*att2
```

```python
# Slicing: chia nhóm sinh viên đầu - giữa - cuối
n = len(sid1)
group_top = sid1[:n//3]
group_mid = sid1[n//3:2*n//3]
group_bot = sid1[2*n//3:]

print("Nhóm đầu:", group_top[:5], "...")   # in vài ID đầu
print("Nhóm giữa:", group_mid[:5], "...")
print("Nhóm cuối:", group_bot[:5], "...")
```

```python
# Phân loại theo tổng điểm (boolean mask)
def classify(total):
    mask_xs = total >= 8.0
    mask_kha = (total >= 6.5) & (total < 8.0)
    mask_tb = (total >= 5.0) & (total < 6.5)
    mask_yeu = total < 5.0
    return mask_xs, mask_kha, mask_tb, mask_yeu

mask_xs1, mask_kha1, mask_tb1, mask_yeu1 = classify(total1)
mask_xs2, mask_kha2, mask_tb2, mask_yeu2 = classify(total2)

print("Phân loại học lực kỳ 1:")
print("Xuất sắc:", np.sum(mask_xs1))
print("Khá:", np.sum(mask_kha1))
print("Trung bình:", np.sum(mask_tb1))
print("Yếu:", np.sum(mask_yeu1))

print("Phân loại học lực kỳ 2:")
print("Xuất sắc:", np.sum(mask_xs2))
print("Khá:", np.sum(mask_kha2))
print("Trung bình:", np.sum(mask_tb2))
print("Yếu:", np.sum(mask_yeu2))

# Báo cáo tiến bộ/giảm sút
diff_total = total2 - total1
progress_ids = sid1[diff_total > 0]
decline_ids = sid1[diff_total < 0]

print("Danh sách sinh viên tiến bộ:", progress_ids[:10], "...")
print("Danh sách sinh viên giảm sút:", decline_ids[:10], "...")
```

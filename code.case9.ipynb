```python
import numpy as np
# đọc dữ liệu từ 2 tệp
file_1="semester1_scores (1).csv"
file_2="semester2_scores.csv"
#đọc dữ liệu:
data_1= np.genfromtxt(file_1,delimiter=",",dtype=float,skip_header=1)
data_2= np.genfromtxt(file_2,delimiter=",",dtype=float,skip_header=1)
# In toàn bộ dữ liệu đọc từ semester1_scores.csv
print(" dữ liệu kì 1 semester1_scores: ")
print(data_1)
# in ra dữ liệu kì 2
print("\n dữ liệu kì 2 semester2_scores: ")
print(data_2)
```

```python
# NHIỆM VỤ 2: TÁCH 4 CỘT THÀNH MẢNG RIÊNG
# Học kỳ 1
student_id_1 = data_1[:, 0].astype(int)   # mã sinh viên
midterm_1    = data_1[:, 1]               # giữa kỳ
final_1      = data_1[:, 2]               # cuối kỳ
att_1        = data_1[:, 3]               # chuyên cần / bài tập

# Học kỳ 2
student_id_2 = data_2[:, 0].astype(int)
midterm_2    = data_2[:, 1]
final_2      = data_2[:, 2]
att_2        = data_2[:, 3]

print("\n=== KIỂM TRA TÁCH CỘT HK1 ===")
print("student_id_1:", student_id_1[:10])
print("midterm_1   :", midterm_1[:10])
print("final_1     :", final_1[:10])
print("att_1       :", att_1[:10])
```

```python
# NHIỆM VỤ 3: TÍNH ĐIỂM TỔNG
# total = 0.3*midterm + 0.5*final + 0.2*attendance
total_1 = 0.3 * midterm_1 + 0.5 * final_1 + 0.2 * att_1
total_2 = 0.3 * midterm_2 + 0.5 * final_2 + 0.2 * att_2
print("\n=== ĐIỂM TỔNG HK1 ===")
print(total_1[:10])
print("\n=== ĐIỂM TỔNG HK2 ===")
print(total_2[:10])
```

```python
#nhiệm vụ 4  Kiểm tra dtype, shape; in 5 giá trị đầu/cuối
# In kiểu dữ liệu và kích thước mảng HK1
print("HK1:")
print("data_1 dtype:", data_1.dtype)   # dtype = kiểu dữ liệu trong mảng (float64)
print("data_1 shape:", data_1.shape)  # shape = (số dòng, số cột)
# In kiểu dữ liệu và kích thước mảng HK2
print("\nHK2:")
print("data_2 dtype:", data_2.dtype)
print("data_2 shape:", data_2.shape)
# In 5 điểm tổng đầu tiên và 5 điểm tổng cuối cùng của HK1
print("\nTotal HK1 - 5 đầu:", total_1[:5])   # [:5] lấy 5 phần tử đầu
print("Total HK1 - 5 cuối:", total_1[-5:])  # [-5:] lấy 5 phần tử cuối
# In 5 điểm tổng đầu tiên và 5 điểm tổng cuối cùng của HK2
print("\nTotal HK2 - 5 đầu:", total_2[:5])
print("Total HK2 - 5 cuối:", total_2[-5:])
```

```python
# NHIỆM VỤ 5 Slicing: nhóm sinh viên đầu – giữa – cuối
n = len(student_id_1)   # n là số sinh viên trong HK1
# Tính vị trí chia 1/3 và 2/3
one_third = n // 3       # n//3: chia lấy phần nguyên (1/3 danh sách)
two_third = 2 * n // 3   # 2/3 danh sách

# --- Nhóm đầu ---
first_group_ids = student_id_1[:one_third]      # lấy từ đầu tới trước 1/3
first_group_total = total_1[:one_third]         # điểm tổng tương ứng

# --- Nhóm giữa ---
middle_group_ids = student_id_1[one_third:two_third]   # lấy từ 1/3 tới trước 2/3
middle_group_total = total_1[one_third:two_third]

# --- Nhóm cuối ---
last_group_ids = student_id_1[two_third:]       # lấy từ 2/3 tới hết
last_group_total = total_1[two_third:]

# In thử số lượng từng nhóm
print("Số SV nhóm đầu:", len(first_group_ids))
print("Số SV nhóm giữa:", len(middle_group_ids))
print("Số SV nhóm cuối:", len(last_group_ids))

# Xem vài SV cho chắc
print("\nNhóm đầu (5 SV đầu):", first_group_ids[:5])
print("Nhóm giữa (5 SV đầu):", middle_group_ids[:5])
print("Nhóm cuối (5 SV đầu):", last_group_ids[:5])
```

```python
# NHIỆM VỤ 6 Boolean mask phân loại: xuất sắc, khá, trung bình, yếu
# Tạo mask boolean cho HK1
excel_1 = total_1 >= 8.5                 # True nếu SV >= 8.5
good_1  = (total_1 >= 7) & (total_1 < 8.5)   # True nếu 7 <= total < 8.5
avg_1   = (total_1 >= 5) & (total_1 < 7)     # True nếu 5 <= total < 7
weak_1  = total_1 < 5                   # True nếu total < 5
# Đếm số SV mỗi loại bằng np.sum(mask)
print("HK1 - Xuất sắc:", np.sum(excel_1))
print("HK1 - Khá:", np.sum(good_1))
print("HK1 - Trung bình:", np.sum(avg_1))
print("HK1 - Yếu:", np.sum(weak_1))
# Lấy danh sách ID theo từng loại
excel_ids_1 = student_id_1[excel_1]   # lấy các id có mask True
good_ids_1  = student_id_1[good_1]
avg_ids_1   = student_id_1[avg_1]
weak_ids_1  = student_id_1[weak_1]
# In thử vài SV cho đẹp
print("\nMột vài SV xuất sắc HK1:", excel_ids_1[:10])
print("Một vài SV yếu HK1:", weak_ids_1[:10])
# Mask cho HK2
excel_2 = total_2 >= 8.5
good_2  = (total_2 >= 7) & (total_2 < 8.5)
avg_2   = (total_2 >= 5) & (total_2 < 7)
weak_2  = total_2 < 5

print("\nHK2 - Xuất sắc:", np.sum(excel_2))
print("HK2 - Khá:", np.sum(good_2))
print("HK2 - Trung bình:", np.sum(avg_2))
print("HK2 - Yếu:", np.sum(weak_2))
```

```python
# NHIỆM VỤ 7
# Broadcasting: diff_total giữa 2 kỳ

# diff_total là mức tăng/giảm điểm tổng của từng sinh viên
# total_2 và total_1 đều là mảng 1 chiều (mỗi phần tử là 1 SV)
diff_total = total_2 - total_1

# abs_diff = độ lớn thay đổi (bỏ dấu âm/dương)
abs_diff = np.abs(diff_total)

# Kiểm tra nhanh 5 SV đầu
print("diff_total 5 SV đầu:", diff_total[:5])
print("abs_diff 5 SV đầu:", abs_diff[:5])

# Đếm số SV tăng/giảm/không đổi
num_improved = np.sum(diff_total > 0)     # True nếu tăng
num_declined = np.sum(diff_total < 0)     # True nếu giảm
num_same     = np.sum(diff_total == 0)    # True nếu không đổi

print("\nSố SV tăng điểm:", num_improved)
print("Số SV giảm điểm:", num_declined)
print("Số SV giữ nguyên:", num_same)
```

```python
# NHIỆM VỤ 8
# Fancy indexing: top-10 theo từng học kỳ

# argsort(total_1) trả về chỉ số sắp xếp total_1 từ thấp -> cao
sorted_idx_1 = np.argsort(total_1)

# Lấy 10 chỉ số cuối (cao nhất), rồi đảo ngược lại để thành cao -> thấp
top10_idx_1 = sorted_idx_1[-10:][::-1]

# Fancy indexing để lấy id và điểm top10 HK1
top10_ids_1 = student_id_1[top10_idx_1]
top10_scores_1 = total_1[top10_idx_1]

print("TOP 10 HK1 (student_id, total):")
print(np.column_stack([top10_ids_1, top10_scores_1]))


# ---------- Làm y hệt cho HK2 ----------
sorted_idx_2 = np.argsort(total_2)
top10_idx_2 = sorted_idx_2[-10:][::-1]

top10_ids_2 = student_id_2[top10_idx_2]
top10_scores_2 = total_2[top10_idx_2]

print("\nTOP 10 HK2 (student_id, total):")
print(np.column_stack([top10_ids_2, top10_scores_2]))
```

```python
# NHIỆM VỤ 9
# Phân tích SV thay đổi mạnh |Δtotal| > 1.5

# big_change_mask là mảng True/False
# True nếu SV thay đổi mạnh hơn 1.5 điểm
big_change_mask = abs_diff > 1.5

# Lọc ra danh sách student_id thay đổi mạnh
big_change_ids = student_id_1[big_change_mask]

# Và mức thay đổi tương ứng
big_change_diff = diff_total[big_change_mask]

print("Số SV thay đổi mạnh (|Δ| > 1.5):", len(big_change_ids))

print("\nMột vài SV thay đổi mạnh (id, diff):")
print(np.column_stack([big_change_ids[:10], big_change_diff[:10]]))
```

```python
# NHIỆM VỤ 10
# Thống kê mô tả theo học kỳ

def stats_1_array(x, name):
    """
    Hàm in thống kê mô tả cho 1 mảng điểm x
    name: tên loại điểm để in cho dễ nhìn
    """
    print(f"--- {name} ---")
    print("Mean   =", np.mean(x))          # trung bình
    print("Std    =", np.std(x))           # độ lệch chuẩn
    print("Min    =", np.min(x))           # nhỏ nhất
    print("Max    =", np.max(x))           # lớn nhất
    print("Median =", np.median(x))        # trung vị
    print()

print("========== THỐNG KÊ HỌC KỲ 1 ==========")
stats_1_array(midterm_1, "Midterm HK1")
stats_1_array(final_1,  "Final HK1")
stats_1_array(att_1,    "Attendance/HW HK1")
stats_1_array(total_1,  "Total HK1")

print(" THỐNG KÊ HỌC KỲ 2 ")
stats_1_array(midterm_2, "Midterm HK2")
stats_1_array(final_2,  "Final HK2")
stats_1_array(att_2,    "Attendance/HW HK2")
stats_1_array(total_2,  "Total HK2")
```

```python

# NHIỆM VỤ 11
# So sánh điểm tổng hai học kỳ
# 1So sánh tổng quan (mean HK2 - mean HK1)
mean_diff = np.mean(total_2) - np.mean(total_1)
print("Mean(HK2) - Mean(HK1) =", mean_diff)

# Sinh viên tiến bộ nhất
best_improve_idx = np.argmax(diff_total)     # chỉ số có diff_total lớn nhất
best_improve_id  = student_id_1[best_improve_idx]
best_improve_val = diff_total[best_improve_idx]

print("\nSV tiến bộ nhất:")
print("ID =", best_improve_id, "| tăng =", best_improve_val)

#  Sinh viên tụt mạnh nhất
worst_drop_idx = np.argmin(diff_total)       # chỉ số có diff_total nhỏ nhất
worst_drop_id  = student_id_1[worst_drop_idx]
worst_drop_val = diff_total[worst_drop_idx]

print("\nSV tụt mạnh nhất:")
print("ID =", worst_drop_id, "| giảm =", worst_drop_val)
```
